# 01 Data Exploration

```python
# Import pandas for working with our F1 datasets
import pandas as pd
```

```python
sessions = pd.read_csv(
    "https://api.openf1.org/v1/sessions?year=2024&csv=true"
)

races = sessions[
    (sessions["session_name"] == "Race") &
    (sessions["session_type"] == "Race")
].copy()

races = races.sort_values("date_start")

print("Number of race sessions:", len(races))

races[["country_name", "location", "date_start", "session_key"]]
```

```python
# Save the 2024 race sessions to a CSV file
races.to_csv("f1_2024_race_sessions.csv", index=False)

# Confirm how many races we saved
print("Race sessions saved:", len(races))
```

```python
import time

# Store the lap data from each race
all_laps = []

# Go through our 24 races one at a time
for _, race in races.iterrows():
    session_key = race["session_key"]
    country = race["country_name"]

    # Download lap times for this race
    url = f"https://api.openf1.org/v1/laps?session_key={session_key}&csv=true"

    try:
        race_laps = pd.read_csv(url)

        # Keep the race name with its lap data
        race_laps["race_name"] = country

        all_laps.append(race_laps)
        print(f"Downloaded: {country} ({len(race_laps)} laps)")

    except Exception as error:
        print(f"Could not download {country}: {error}")

    # Small pause between requests
    time.sleep(1)

# Combine the downloaded races into one dataset
if all_laps:
    laps_2024 = pd.concat(all_laps, ignore_index=True)
    print("Total laps downloaded:", len(laps_2024))
else:
    print("No lap data downloaded.")
```

```python
# Save all downloaded 2024 lap records
laps_2024.to_csv("f1_2024_all_laps.csv", index=False)

# Confirm the file was saved
print("Saved", len(laps_2024), "lap records!")
```

```python
# Download the list of 2023 F1 sessions
sessions_2023 = pd.read_csv(
    "https://api.openf1.org/v1/sessions?year=2023&csv=true"
)

# Keep only the main Grand Prix races
races_2023 = sessions_2023[
    (sessions_2023["session_name"] == "Race") &
    (sessions_2023["session_type"] == "Race")
].copy()

# Show the number of races found
print("2023 race sessions found:", len(races_2023))
```

```python
# Store lap data from the 2023 races
all_laps_2023 = []

# Download laps for each race
for _, race in races_2023.iterrows():
    session_key = race["session_key"]
    race_name = race["location"]

    # Create the download link
    url = f"https://api.openf1.org/v1/laps?session_key={session_key}&csv=true"

    try:
        # Read the race's lap data
        race_laps = pd.read_csv(url)

        # Keep the race location for identification
        race_laps["race_name"] = race_name

        all_laps_2023.append(race_laps)
        print(f"Downloaded: {race_name} ({len(race_laps)} laps)")

    except Exception as error:
        print(f"Failed: {race_name} - {error}")

    # Pause briefly between downloads
    time.sleep(1)

# Combine all successful downloads
if all_laps_2023:
    laps_2023 = pd.concat(all_laps_2023, ignore_index=True)
    print("Total 2023 laps:", len(laps_2023))
else:
    print("No lap data downloaded.")
```

```python
# Save all downloaded 2023 lap records
laps_2023.to_csv("f1_2023_all_laps.csv", index=False)

# Confirm how many records were saved
print("Saved", len(laps_2023), "lap records!")
```

```python
# Store the lap data from 2025
all_laps_2025 = []

# Go through each 2025 race
for _, race in races_2025.iterrows():
    session_key = race["session_key"]
    race_name = race["location"]

    # Download lap times for this race
    url = f"https://api.openf1.org/v1/laps?session_key={session_key}&csv=true"

    try:
        # Read the lap data
        race_laps = pd.read_csv(url)

        # Add the race location
        race_laps["race_name"] = race_name

        # Store the downloaded records
        all_laps_2025.append(race_laps)

        print(f"Downloaded: {race_name} ({len(race_laps)} laps)")

    except Exception as error:
        print(f"Failed: {race_name} - {error}")

    # Pause briefly between requests
    time.sleep(1)

# Combine all successfully downloaded races
if all_laps_2025:
    laps_2025 = pd.concat(all_laps_2025, ignore_index=True)
    print("Total 2025 laps:", len(laps_2025))
else:
    print("No lap data downloaded.")
```

```python
import time

# Find the two races that failed
missing_races = races_2025[
    races_2025["location"].isin(["Spielberg", "Silverstone"])
]

# Download only their missing tyre data
for _, race in missing_races.iterrows():
    session_key = race["session_key"]
    race_name = race["location"]

    # Create the API link
    url = f"https://api.openf1.org/v1/stints?session_key={session_key}&csv=true"

    try:
        # Download the missing tyre stints
        missing_stints = pd.read_csv(url)

        # Add the race location
        missing_stints["race_name"] = race_name

        # Add the missing records to our existing dataset
        stints_2025 = pd.concat(
            [stints_2025, missing_stints],
            ignore_index=True
        )

        print(f"Recovered: {race_name} ({len(missing_stints)} stints)")

    except Exception as error:
        print(f"Still unavailable: {race_name} - {error}")

    # Give the API a longer break
    time.sleep(5)

# Check the updated totals
print("Total 2025 tyre stints:", len(stints_2025))
print("Races covered:", stints_2025["session_key"].nunique())
```

```python
# Save the repaired 2025 tyre dataset
stints_2025.to_csv("f1_2025_stints.csv", index=False)

# Confirm the file was saved
print("Saved", len(stints_2025), "tyre stints!")
```

```python
# Calculate the median lap duration for each race session
race_medians = (
    normal_pace_laps
    .groupby(["session_key", "race_name"])["lap_duration"]
    .median()
    .reset_index(name="median_lap_duration")
)

# Check how many races we have and preview the results
print("Race sessions:", len(race_medians))
print(race_medians.head(10).to_string(index=False))
```

```python
# Add each race's median lap time to its individual laps
normal_pace_laps["race_median"] = (
    normal_pace_laps.groupby("session_key")["lap_duration"]
    .transform("median")
)

# Calculate how slow each lap is relative to its race
normal_pace_laps["pace_ratio"] = (
    normal_pace_laps["lap_duration"] /
    normal_pace_laps["race_median"]
)

# Examine the distribution of relative lap pace
print(
    normal_pace_laps["pace_ratio"]
    .describe(percentiles=[0.90, 0.95, 0.99])
    .round(3)
)
```

```python
# Identify laps taking at least twice the race's median time
extreme_relative_laps = normal_pace_laps[
    normal_pace_laps["pace_ratio"] >= 2.0
]

# Count affected laps and race sessions
print("Extreme relative laps:", len(extreme_relative_laps))
print("Race sessions affected:", extreme_relative_laps["session_key"].nunique())
```
